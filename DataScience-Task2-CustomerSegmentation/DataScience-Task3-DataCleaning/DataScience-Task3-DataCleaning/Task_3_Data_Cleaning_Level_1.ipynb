## Missing Data Handling

The missing values were found in three columns: Age, Cabin, and Embarked.

The following strategies are selected:

| Column | Missing Values | Strategy | Reason |
|---|---:|---|---|
| Age | 177 | Median imputation | Age is numerical and the median is less affected by extreme values |
| Cabin | 687 | Drop column | A very large proportion of values are missing, making reliable imputation impractical |
| Embarked | 2 | Mode imputation | Embarked is categorical and only two values are missing |

These decisions are made based on the data type, amount of missing data,
and the meaning of each variable.

```python
age_median = df['Age'].median()

print("Median Age:", age_median)
```

### Observation — Age

The median age is 28.0 years. Since Age is a numerical variable with
missing values, median imputation is selected.

The median is preferred over the mean because it is more robust to
extreme age values and helps preserve the overall distribution of the
dataset.

```python
df['Age'] = df['Age'].fillna(age_median)

print("Missing Age values after imputation:", df['Age'].isnull().sum())
```

```python
print("Embarked value counts before imputation:")
print(df['Embarked'].value_counts(dropna=False))
```

### Observation — Embarked

The Embarked column is a categorical variable with only 2 missing values.

Since only a very small number of records are missing and the column is
categorical, the most frequent category (mode) is an appropriate
replacement strategy.

```python
embarked_mode = df['Embarked'].mode()[0]

print("Most frequent Embarked value:", embarked_mode)
```

```python
df['Embarked'] = df['Embarked'].fillna(embarked_mode)

print("Missing Embarked values after imputation:",
      df['Embarked'].isnull().sum())
```

### Observation — Embarked Imputation

The mode of the Embarked column is "S". The 2 missing values were replaced
with this most frequent category.

This approach is appropriate because Embarked is categorical and only a
very small proportion of its values were missing.

### Observation — Cabin Missing Data

The Cabin column contains 687 missing values out of 891 records,
representing approximately 77% of the column.

Because such a large proportion of the values are missing, mean, median,
mode, or forward-fill imputation would not provide reliable information.

Therefore, the Cabin column will be removed from the dataset. This prevents
the introduction of large amounts of artificial or misleading data.

```python
df = df.drop(columns=['Cabin'])

print("Cabin column removed successfully.")
print("New dataset shape:", df.shape)
```

```python
print("Missing values after handling:")
print(df.isnull().sum())
```

```python
numeric_columns = ['Age', 'SibSp', 'Parch', 'Fare']

outlier_report = []

for col in numeric_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]

    outlier_report.append({
        'Column': col,
        'Q1': Q1,
        'Q3': Q3,
        'IQR': IQR,
        'Lower Bound': lower_bound,
        'Upper Bound': upper_bound,
        'Outlier Count': len(outliers)
    })

outlier_report = pd.DataFrame(outlier_report)

outlier_report
```

### Observation — IQR Outlier Detection

The IQR method was used to identify potential outliers in the main
continuous and count-based numerical variables.

An observation is considered a potential outlier when it falls below
Q1 - 1.5 × IQR or above Q3 + 1.5 × IQR.

The identified values will not automatically be removed because an
outlier may represent a valid real-world observation rather than an error.

### Outlier Treatment Decision

The detected outliers will be retained because extreme values in this
dataset can represent genuine passengers or valid ticket fares.

For example, a high Fare value may correspond to a premium ticket, while
higher SibSp or Parch values may represent passengers travelling with
larger families.

Removing these records could result in loss of useful information and
could introduce bias.

Therefore, no valid observations will be removed or capped solely because
they are statistical outliers.

```python
print("Rows after outlier analysis:", len(df))
```

### Observation — Final Outlier Decision

The outlier analysis was performed using the IQR method. Potential
outliers were identified, but no rows were removed because the extreme
values can represent legitimate passenger characteristics or fares.

The dataset therefore retains all 891 valid records.
